````python
# Plot the percentage of missing values in each incomplete column.
missing_cols = (
    profile.loc[profile["Missing Percent"] > 0]
    .sort_values("Missing Percent", ascending=True)
)

fig = px.bar(
    missing_cols,
    x="Missing Percent",
    y="Column",
    orientation="h",
    text="Missing Percent",
    title="<b>Missing Data by Column</b><br><span style='font-size: 12px; color: #6B7280;'>Percentage of null values per field</span>",
    template="plotly_white",
    height=380,
)

fig.update_traces(
    marker_color="#4F46E5",
    marker_line_width=0,
    texttemplate="%{text:.1f}%",
    textposition="outside",
    cliponaxis=False,
    hovertemplate="<b>%{y}</b><br>Missing: %{x:.1f}%<extra></extra>",
)

fig.update_layout(
    xaxis={
        "title": "Missing (%)",
        "range": [0, 110],
        "showgrid": True,
        "gridcolor": "#F3F4F6",
        "zeroline": False,
        "ticksuffix": "%",
    },
    yaxis={
        "title": "",
        "tickfont": {"size": 12, "color": "#374151"},
    },
    plot_bgcolor="white",
    paper_bgcolor="white",
    font={"family": "system-ui, -apple-system, sans-serif", "color": "#111827"},
    margin={"l": 20, "r": 40, "t": 60, "b": 40},
)

fig.show()
````

````python
# Plot the share of postings at each career level.
career_levels = (
    jobs["Career Level"]
    .fillna("Missing")
    .value_counts(normalize=True, dropna=False)
    .rename_axis("Career Level")
    .reset_index(name="Postings")
)

fig = px.bar(
    career_levels,
    x="Career Level",
    y="Postings",
    text="Postings",
    title="<b>Job Postings by Career Level</b>",
    template="plotly_white",
)

fig.update_traces(
    texttemplate="%{text:.1%}",
    textposition="outside",
    marker_color="#4F46E5",
    hovertemplate="<b>%{x}</b><br>Share: %{y:.1%}<extra></extra>",
)

fig.show()
````

````python
# Plot the 15 job categories with the most postings.
top_categories = (
    jobs["Job Category"]
    .fillna("Missing")
    .value_counts()
    .head(15)
    .rename_axis("Job Category")
    .reset_index(name="Postings")
    .sort_values("Postings")
)

fig = px.bar(
    top_categories,
    x="Postings",
    y="Job Category",
    text="Postings",
    orientation="h",
    title="<b>Top 15 Job Categories</b>",
    template="plotly_white",
)  

fig.show()
````

````python
# Measure missing, unique, and duplicate business titles.
title_quality = (
    jobs["Business Title"]
    .fillna("")
    .str.strip()
    .replace("", np.nan)
)

print(f"Missing or blank titles: {title_quality.isna().sum():,}")
print(f"Unique titles: {title_quality.nunique():,}")
print(f"Duplicate titles: {title_quality.duplicated().sum():,}")
````

````python
# Summarize the lengths of titles, descriptions, and requirements.
text_quality = pd.DataFrame(
    {
        "title_characters": jobs["Business Title"].fillna("").str.len(),
        "description_words": jobs["Job Description"].fillna("").str.split().str.len(),
        "requirements_words": (
            jobs["Minimum Qual Requirements"].fillna("").str.split().str.len()
        ),
    }
)

text_quality.describe()
````

````python
# Summarize annual salary ranges and their midpoints.
annual_salaries = jobs.loc[
    jobs["Salary Frequency"].eq("Annual"),
    ["Salary Range From", "Salary Range To", "Career Level"],
].copy()

annual_salaries["salary_from"] = pd.to_numeric(
    annual_salaries["Salary Range From"], errors="coerce"
)

annual_salaries["salary_to"] = pd.to_numeric(
    annual_salaries["Salary Range To"], errors="coerce"
)

annual_salaries["salary_midpoint"] = (
    annual_salaries["salary_from"] + annual_salaries["salary_to"]
) / 2

annual_salaries[["salary_from", "salary_to", "salary_midpoint"]].describe()
````

````python
# Plot the number of job postings published each month.
eda = jobs.copy()
eda["posting_date"] = pd.to_datetime(eda["Posting Date"], errors="coerce")

print(f"Invalid or missing posting dates: {eda['posting_date'].isna().sum():,}")

postings_by_month = (
    eda.dropna(subset=["posting_date"])
    .assign(month=lambda frame: frame["posting_date"].dt.to_period("M").astype(str))
    .groupby("month")
    .size()
    .reset_index(name="Postings")
)

fig = px.line(
    postings_by_month,
    x="month",
    y="Postings",
    markers=True,
    title="Job Postings by Month",
)

fig.show()
````

The dataset has **2,362 rows** and **1,228 unique job IDs**. Most repeated IDs are internal and external versions of the same job. For example, job `540190` appears twice as `Assistant District Attorney - Economic Crimes Bureau`, once for each posting type. Before modeling, combine these records into one job and keep the posting types as metadata.

There are also **10 exact duplicate rows**. For example Job `776352`, `Associate Water Use Inspector`, contains the same internal record twice. These copies can be removed safely.


For the mvp model we can use `Business Title` as the embedding text since it is complete and relevant to our initial modeling.

We can keep `Civil Service Title` for clustering. For example, `Project Manager`, `Engineering Auditor`, and `Senior Project Administrator` all use the civil service title `ASSOCIATE PROJECT MANAGER`, which can help group related roles.

We can use `Job Category` to color and filter the map Some categories are broad. For example, `Executive Assistant` appears under `Technology, Data & Innovation`.

We can use `Career Level` to label seniority and direct possible career paths. The data is heavily skewed towards `Experienced(None-manager)` roles

- Experienced non-manager: **73.2%**
- Entry-level: **11.6%**
- Manager: **10.2%**
- Student: **2.8%**
- Executive: **2.2%**

The labels are useful but broad. For example, `Senior Project Manager` is still labeled `Experienced (non-manager)`. Keep seniority terms from the title when cleaning the text.

### Text fields for later experiments

`Job Description` is complete, and `Minimum Qual Requirements` is available for **98.5%** of rows. These fields can support a later model. Start with titles so repeated agency language does not affect the first set of clusters.

Keep `Preferred Skills` optional because it is missing from **50.8%** of rows. For example, the `Project Manager` posting includes construction-management skills, while the `Public Health Inspector` posting has no preferred-skills value.

### Preprocessing handoff

1. Remove the 10 exact duplicate rows.
2. Create one modeling row per `Job ID` and preserve all posting types.
3. Create `normalized_title` by fixing capitalization, tabs, repeated spaces, and minor punctuation differences.
4. Preserve terms such as `Junior`, `Senior`, `Lead`, `Manager`, `Director`, and `Level 1`.
5. Create `seniority_rank` from `Career Level`: `Student = 0`, `Entry-Level = 1`, `Experienced (non-manager) = 2`, `Manager = 3`, and `Executive = 4`.
6. Use `normalized_title` for the first embeddings. Keep the original title, civil service title, category, career level, and posting type for labels, filters, and validation.

For example:

```text
Assistant District Attorney -     Economic Crimes Bureau
-> assistant district attorney - economic crimes bureau
```

Leave salary out of the first embedding. The file mixes annual and hourly salaries, and some annual records have a minimum salary of zero. For example, `Systems Developer` has an annual range of `$0-$61,194`, while `Plumber, Bureau of Operations` has an hourly rate of `$64`. Salary needs separate validation before we use it as a career signal.

After preprocessing, validate the row count and required fields. Then generate the title embeddings and reduce them to three dimensions.
